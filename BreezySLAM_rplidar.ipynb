```python
import numpy as np

# Начало координат в BreezySLAM - верхний левый угол, как в массивах numpy, в браузере

target_x = 5500
target_y = 5500
robot_x = 5000
robot_y = 5000
robot_angle = -20

dx = target_x - robot_x
dy = target_y - robot_y
print(f"dx: {dx}, dy: {dy}")
target_angle_rad = np.arctan2(dy, dx)
target_angle_deg = np.degrees(target_angle_rad)
print(f"target angle: {target_angle_deg}, robot_angle: {robot_angle}")

angle_diff = target_angle_deg - robot_angle

print(f"Разница угла: {angle_diff}")

turn_direction = "right" if angle_diff > 0 else "left"
turn_angle = abs(angle_diff)
```
