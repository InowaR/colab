```python
import numpy as np

def diff_central(func, x, dx=0.01):
    return (func(x + dx) - func(x - dx)) / (2 * dx)

X = np.random.uniform(-10, 10, size=(100, 3))
y = (X[:, 0] + X[:, 1]) * X[:, 2]

X_poly = np.column_stack([
    (X[:, 0] + X[:, 1]) * X[:, 2],
    (X[:, 0] + X[:, 2]) * X[:, 1],
    (X[:, 1] + X[:, 2]) * X[:, 0]
])

a1, a2, a3, b = 0.0, 0.0, 0.0, 0.0
lr = 0.0001

loss = lambda: np.mean((
    a1 * X_poly[:, 0] +
    a2 * X_poly[:, 1] +
    a3 * X_poly[:, 2] +
    b - y
) ** 2)

loss_a1 = lambda a1_val: np.mean((
    a1_val * X_poly[:, 0] +
    a2 * X_poly[:, 1] +
    a3 * X_poly[:, 2] +
    b - y
) ** 2)

loss_a2 = lambda a2_val: np.mean((
    a1 * X_poly[:, 0] +
    a2_val * X_poly[:, 1] +
    a3 * X_poly[:, 2] +
    b - y
) ** 2)

loss_a3 = lambda a3_val: np.mean((
    a1 * X_poly[:, 0] +
    a2 * X_poly[:, 1] +
    a3_val * X_poly[:, 2] +
    b - y
) ** 2)

loss_b = lambda b_val: np.mean((
    a1 * X_poly[:, 0] +
    a2 * X_poly[:, 1] +
    a3 * X_poly[:, 2] +
    b_val - y
) ** 2)

prev_loss = float('inf')
tolerance = 1e-8
max_iterations = 1000

for i in range(max_iterations):
    current_loss = loss()
    a1 -= lr * diff_central(loss_a1, a1)
    a2 -= lr * diff_central(loss_a2, a2)
    a3 -= lr * diff_central(loss_a3, a3)
    b -= lr * diff_central(loss_b, b)
    if abs(prev_loss - current_loss) < tolerance:
        print(f"Сошлось за {i} итераций")
        break
    prev_loss = current_loss

print(f"\ny = {a1:.3f}(x₁+x₂)x₃ + {a2:.3f}(x₁+x₃)x₂ + {a3:.3f}(x₂+x₃)x₁ + {b:.3f}")
print(f"Ошибка (на исходных данных): {loss():.6f}")

new_sample = np.array([3, -3, 2])
new_poly = np.array([
    (new_sample[0] + new_sample[1]) * new_sample[2],
    (new_sample[0] + new_sample[2]) * new_sample[1],
    (new_sample[1] + new_sample[2]) * new_sample[0]
])
prediction = a1 * new_poly[0] + a2 * new_poly[1] + a3 * new_poly[2] + b
print(f"Предсказание для [{new_sample[0]}, {new_sample[1]}, {new_sample[2]}]: {prediction:.3f}")
```
