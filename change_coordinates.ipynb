```python
lst = [[0, 0], [1, 1], [2, 2], [3, 3]]

for i, point in enumerate(lst):
    print(f"Обрабатываем индекс {i}, точка {point}")
    # Меняем ВСЕ элементы списка на новые
    lst[:] = [[x + 10, y + 10] for x, y in lst]
    # В реальном коде здесь легко получить непредсказуемое поведение
    import time
    time.sleep(1)
```
